Calculate Variance.
Calculate Standard Deviation.
Display the results.
Explain what the standard deviation indicates about the spread of the data.

The standard deviation measures the spread or variability of sales values around the mean. A higher standard deviation indicates that sales values are widely spread from the average, while a lower standard deviation indicates that most sales values are closer to the average.

```python
import numpy as np;
import pandas as pd;
dataset=pd.read_csv('ecommerce_statistics_eda_dataset.csv')
print("-Variance and Standard Deviation-")
print("Variance:",dataset['Sales'].var())
print("Standard Deviation:",dataset['Sales'].std())
```
