Using a suitable numerical dataset:

Calculate the Mean.

*Brief Explanation*
1. Mean is showing averages value of sales data.
2. Meadian is showing middle value of sales data is data is in ordered.
3. Mode is showing which sales values is with highest frequancy.



```python
import numpy as np;
import pandas as pd;
dataset=pd.read_csv('ecommerce_statistics_eda_dataset.csv')
print("-Mean, Medain and Mode of Sales column-")
print("Mean:",dataset['Sales'].mean())
print("Median:",dataset['Sales'].median())
print("Mode:",dataset['Sales'].mode()[0])
```
