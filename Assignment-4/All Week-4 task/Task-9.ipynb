Task 9: Business Recommendations (5 Marks)
Based on your EDA and findings:
Provide at least 2 data-driven business recommendations.
Explain how your recommendations are supported by the data

***Recommendation 1: Improve Sales Performance***

Finding: Sales and Quantity have a positive correlation of 0.41.

Recommendation: The business should analyze its best-selling products and consider maintaining sufficient inventory for products with higher sales volumes. It can also investigate whether promotions or product bundles help increase the quantity sold and overall sales revenue.

Recommendation 2: Improve Customer Satisfaction

Finding: Sales and Rating have a very weak negative correlation of -0.08.

Recommendation: The business should not assume that higher sales automatically lead to higher customer satisfaction. It should analyze customer feedback, product quality, delivery experiences, and complaints to identify opportunities to improve customer ratings.

My recommendations are supported by the sales and customer rating data. Products with the highest sales should be kept sufficiently available in inventory to meet customer demand and avoid stock shortages. By comparing sales figures with customer ratings, we can identify products that sell well but receive low ratings, indicating potential customer dissatisfaction. This analysis helps Adventure Works improve product quality, maintain customer satisfaction, and make better inventory decisions.
