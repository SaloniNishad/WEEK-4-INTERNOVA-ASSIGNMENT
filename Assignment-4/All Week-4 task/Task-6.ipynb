```python
# Count plot
import pandas as pd;
import matplotlib.pyplot as plt;
import seaborn as sns;
dataset=pd.read_csv('ecommerce_statistics_eda_dataset.csv')
sns.countplot(data=dataset, x="Order_Status")
plt.title("Number of Orders by Status")
plt.xlabel("Order Status")
plt.ylabel("Number of Orders")
plt.show()
```

```python
# Box plot
sns.boxplot(data=dataset, y="Sales")

plt.title("Sales Distribution and Outliers")
plt.ylabel("Sales")
plt.show()
```

```python
# Heatmap.
correlation = dataset[["Sales", "Quantity", "Rating"]].corr()
sns.heatmap(correlation, annot=True)

plt.title("Correlation Heatmap")
plt.show()
```

```python
# pair plot
sns.pairplot(
    dataset[["Sales", "Quantity", "Rating"]]
)

plt.show()
```
