##Correlation
The correlation analysis was performed between Sales and Quantity to determine whether there is a relationship between the amount sold and the quantity purchased

```python
import numpy as np
import pandas as pd

dataset = pd.read_csv('ecommerce_statistics_eda_dataset.csv')
correlation = dataset["Sales"].corr(dataset["Quantity"])
print("Correlation between Sales and Quantity:", correlation)
```

##Probability
What is the probability that a randomly selected order is Delivered?
The probability represents the likelihood of a randomly selected order being delivered. The probability is calculated by dividing the number of delivered orders by the total number of orders.

```python
total_orders = len(dataset)

delivered_orders = (dataset["Order_Status"] == "Delivered").sum()

probability = delivered_orders / total_orders

print("Total Orders:", total_orders)
print("Delivered Orders:", delivered_orders)
print("Probability of Delivered Order:", probability)
print(f"Probability : {(probability * 100):.2f}%")
```
