```python
import matplotlib.pyplot as plt;
import pandas as pd;
dataset=pd.read_csv('ecommerce_statistics_eda_dataset.csv')
```

```python
# Line Chart
# Explanation:This line chart shows the variation in sales across different orders. It helps identify the changes and trends in sales values from one order to another. 
x=dataset['Order_ID']
y=dataset['Sales']
plt.plot(x,y,color='SkyBlue')
plt.title("Sales by Order")
plt.xlabel("Order_ID")
plt.ylabel("Sales")
plt.show()
```

```python
# Bar Chart
# Explanation:The bar chart shows the total sales for different product categories. Each bar represents a category, and the height of the bar indicates its sales value. It helps compare sales performance across categories and identify which category has the highest or lowest sales.
x=dataset['Category']
y=dataset['Sales']
plt.bar(x, y, color = 'orange', label='Category Vs Sales')
plt.title("Sales By Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

```python
# Pie Chart
# Explanation:The pie chart shows the percentage distribution of total sales across different regions. Each slice represents a region, and the size of the slice indicates its contribution to overall sales. It helps understand which region contributes the most to total sales.
region_sales = dataset.groupby("Region")["Sales"].sum()

plt.pie(
    region_sales.values,
    labels=region_sales.index,
    autopct="%1.1f%%"
)

plt.title("Sales Distribution by Region")
plt.show()
```

```python
# Histogram chart
# Explanation:The histogram shows the distribution of sales values across different sales ranges. The bars represent the number of sales records falling within each range. It helps understand the spread and frequency of sales and identify whether most sales are concentrated in a particular range.
x=dataset['Sales']
plt.hist(x, bins = 10, color='purple', edgecolor = 'black')
plt.title("Sales Distribution")
plt.xlabel("Sales Range")
plt.ylabel("Number of Sales")
plt.show()
```

```python
# Scatter plot
# Explanation:The scatter plot shows the relationship between Quantity and Sales. Each point represents a data record, with Quantity on the x-axis and Sales on the y-axis. It helps identify whether higher quantities are associated with higher sales and can also reveal unusual data points or outliers.

plt.scatter(dataset['Quantity'],dataset['Sales'], color='Green',marker='*', label='Sales Vs Quantity')
plt.legend()
plt.title("Sales Vs Qunatity")
plt.xlabel("Quantity")
plt.ylabel("Sales")
plt.grid()
plt.show()
```
