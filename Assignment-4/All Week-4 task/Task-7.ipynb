```python
import numpy as np
import pandas as pd

dataset = pd.read_csv('ecommerce_statistics_eda_dataset.csv')
```

```python
# Check the number of rows and columns.

print("Number of columns and rows")
print(dataset.info())
```

```python
# Display column names.
print("Columns Name:")
print(dataset.columns)
```

```python
# Check data types.
print("Data type")
print("Sales:",(dataset['Sales']).dtype)
print("Order_ID:",(dataset['Order_ID']).dtype)
print("Category:",(dataset['Category']).dtype)
```

```python
# Generate a statistical summary.
print("Statistical summary of dataset:")
print(dataset.describe())
```

```python
print("Dataset bofore cleaning")
print(dataset)
```

```python

# Data Cleaning
print(dataset.isnull().sum())  #finding missing values
dataset['Region']=dataset['Region'].fillna(dataset['Region'].mode()[0], inplace = True)
dataset['Rating']=dataset['Rating'].fillna(dataset['Rating'].mean(), inplace = True)
print(dataset.duplicated().sum()) #finding Duplicate Value
print(dataset[dataset.duplicated()])
dataset = dataset.drop_duplicates()
# Check the dataset for incorrect or inconsistent values.
print(dataset['Category'].unique())
print(dataset['Payment_Method'].unique())
print(dataset['Order_Status'].unique())
print(dataset['Region'].unique())
dataset['Category'] = dataset['Category'].replace({
    'electronics': 'Electronics',
    ' ELECTRONICS ': 'Electronics'
})
```

```python
print("Dataset after cleaning")
print(dataset)
```
